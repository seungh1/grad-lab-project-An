```python
import numpy as np
import matplotlib.pyplot as plt
```

The difference between a code and a markdown cell is the purpose of each cell. A code cell can run code (python for Google collab) while markdown cell is for displaying text. If a code cell is run and a variable in the code cell is changed after, the cell would need to be rerun to update the variable.

```python
image = np.zeros((128, 128))
image[40:88, 40:88] = 1.0

plt.imshow(image, cmap="gray", vmin=0, vmax=1)
plt.title("Synthetic image")
plt.colorbar(label="Pixel intensity")
plt.show()
```

```python
print(f"Image dimension: {image.shape}")
print(f"Minimum Intensity: {image.min()}")
print(f"Minimum Intensity: {image.max()}")
print(f"Mean Intensity: {image.mean()}")
```

```python
image2 = np.zeros((256, 256))
image2[1:100, 1:100] = 1.0

plt.subplot(121)
plt.imshow(image, cmap="gray", vmin=0, vmax=1)
plt.title("Synthetic image")

plt.subplot(122)
plt.imshow(image2, cmap="gray", vmin=0, vmax=1)
plt.title("Synthetic image Version 2")
plt.show()

print(f"Mean intensity of first image: {image.mean()}")
print(f"Mean intensity of second image: {image2.mean()}")
```

The mean intensity of the second image didn't change drastically compared to the mean intensity of the first image.

```python
rng = np.random.default_rng(42)
noisy_image = image + rng.normal(0, 0.1, image.shape)
noisy_image2 = image + rng.normal(0, 0.3, image.shape)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(131)
plt.imshow(image, cmap="gray", vmin=0, vmax=1)
plt.title("Original Image")

plt.subplot(132)
plt.imshow(noisy_image, cmap="gray", vmin=0, vmax=1)
plt.title("Noisy Image (std=0.1)")

plt.subplot(133)
plt.imshow(noisy_image2, cmap="gray", vmin=0, vmax=1)
plt.title("Noisy Image (std=0.3)")

plt.tight_layout()
plt.show()
```

Adding more noise will absolutely affect the visibility of the square. Although this is a simple example with a very visible square, a signal with a low SNR may not be detectable with increased noise levels.

Using the same intensity display is useful for comparison as it allows one to use the same baseline metrics. For this example, one can get the mean intensity of white square to compare how much the noise has affected the image.
